```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from IPython.display import display

print("All libraries imported successfully!")
```

```python

data = {
    "Student_ID": [101, 102, 103, 101, 104, 102, 105, 106, 106, 107],
    "Name": [
        "Rahul", "Priya", "Aman", "Rahul",
        "Simran", "Priya", "Arjun", "Neha",
        "Neha", "Karan"
    ],
    "Age": [20, 21, 22, 20, 20, 21, 23, 22, 22, 21],
    "Course": [
        "BCA", "BCA", "BBA", "BCA",
        "BCA", "BCA", "MCA", "BBA",
        "BBA", "MCA"
    ],
    "City": [
        "Delhi", "Mumbai", "Chandigarh", "Delhi",
        "Ludhiana", "Mumbai", "Delhi", "Amritsar",
        "Amritsar", "Punjab"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
display(df.describe(include="all"))
```

```python

duplicates = df[df.duplicated(keep="first")]

print("Total Duplicate Records:", len(duplicates))

print("\nDuplicate Records:")
display(duplicates)
```

```python

duplicate_count = df.duplicated().sum()
unique_count = len(df) - duplicate_count

plt.figure(figsize=(7, 4))

plt.bar(
    ["Duplicate Records", "Unique Records"],
    [duplicate_count, unique_count],
    color=["red", "green"]
)

plt.title("Duplicate Data Analysis")
plt.ylabel("Number of Records")
plt.show()
```

```python

cleaned_df = df.drop_duplicates(
    keep="first"
).reset_index(drop=True)

print("Cleaned Dataset:")
display(cleaned_df)

print("Original Rows:", len(df))
print("Cleaned Rows:", len(cleaned_df))
print("Removed Duplicates:", len(df) - len(cleaned_df))
```
